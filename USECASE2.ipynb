```python
from sentence_transformers import SentenceTransformer
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM
import numpy as np
documents = [
    "Python is a programming language used for web development, data science, and artificial intelligence.",

    "Artificial Intelligence is the technology used to create systems that can perform tasks requiring human intelligence.",

    "Machine Learning is a branch of Artificial Intelligence where computers learn patterns from data.",

    "RAG stands for Retrieval-Augmented Generation. It retrieves relevant information from a knowledge base and uses it to generate an answer.",

    "Google Colab is an online platform where users can write and run Python programs in a browser.",

    "A chatbot is a computer program that communicates with users using natural language."
]
embedding_model = SentenceTransformer("all-MiniLM-L6-v2")
document_embeddings = embedding_model.encode(documents)
model_name = "google/flan-t5-small"
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForSeq2SeqLM.from_pretrained(model_name)
def rag_chatbot(question):
    question_embedding = embedding_model.encode([question])[0]
    similarity = np.dot(document_embeddings, question_embedding
    best_index = np.argmax(similarity)
    retrieved_document = documents[best_index]
    prompt = f"""
Use the following information to answer the question.
Information:
{retrieved_document}
Question:
{question}
Answer:
"""

    # Convert prompt into tokens
    inputs = tokenizer(
        prompt,
        return_tensors="pt"
    )

    # Generate answer
    outputs = model.generate(
        **inputs,
        max_new_tokens=50
    )

    # Convert tokens to text
    answer = tokenizer.decode(
        outputs[0],
        skip_special_tokens=True
    )

    return retrieved_document, answer


# -------------------------------
# 5. Start Chatbot
# -------------------------------

print("====================================")
print("          SIMPLE RAG CHATBOT")
print("====================================")
print("Ask questions about the knowledge base.")
print("Type 'bye' to exit.\n")

while True:

    question = input("You: ")

    if question.lower() == "bye":
        print("Bot: Goodbye!")
        break

    retrieved, answer = rag_chatbot(question)

    print("\nRetrieved Information:")
    print(retrieved)

    print("\nBot:", answer)
    print()
```
